```python
from collections import deque

#graph - defaultdict consist nodes and neighbours.

def bfs(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)
    
    while queue:
        current = queue.popleft()   # take the OLDEST node in line
        print(current)
        
        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)   # add new nodes to the BACK
```
