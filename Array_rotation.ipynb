```python
def rotate_right(arr, k):
    n = len(arr)
    k = k % n  # handle k larger than array length
    return arr[-k:] + arr[:-k]

arr = [1, 2, 3, 4, 5, 6, 7]
print(rotate_right(arr, 3))  # [5, 6, 7, 1, 2, 3, 4]
```

```python
def rotate_right_inplace(arr, k):
    n = len(arr)
    k = k % n
    
    def reverse(sub_arr, start, end):
        while start < end:
            sub_arr[start], sub_arr[end] = sub_arr[end], sub_arr[start]
            start += 1
            end -= 1
    
    reverse(arr, 0, n - 1) ## reversing whole array- 7 to 1)
    reverse(arr, 0, k - 1) ## reversing K elements- 0-2(3-1))
    reverse(arr, k, n - 1) ## reversing rest array - 3 to 6(7-1))

arr = [1, 2, 3, 4, 5, 6, 7]
rotate_right_inplace(arr, 3)
print(arr)  # [5, 6, 7, 1, 2, 3, 4]
```

```python
def rotate_left(arr, k):
    n = len(arr)
    k = k % n
    return arr[k:] + arr[:k]

print(rotate_left([1,2,3,4,5,6,7], 3))
```
