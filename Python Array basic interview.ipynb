```python
#using sorted() with reverse
```

```python
arr = [4, 1, 9, 7, 3, 9, 2]
```

```python
top2 = sorted(set(arr), reverse = True)[:2]  ## using set to ignore duplicate
print(top2)
```

```python
top2 = sorted((arr), reverse = True)[:2]
print(top2)
```

```python
#using heapq algorithm
```

```python
import heapq

arr = [4, 1, 9, 7, 3, 9, 2]
top2 = heapq.nlargest(2, arr)
print(top2)  # [9, 9]
```

```python
## two largest o(n)- single pass
```

```python
def two_largest(arr):
    first = second = float('-inf')
    for num in arr:
        if num > first:
            first, second = num, first
        elif num > second:
            second = num
    return first, second

print(two_largest([4, 1, 9, 7, 3, 9, 2]))  # (9, 9)
```

```python
def max_num(arr):
    max = arr[0]  
    for num in arr[1:]:
        if num > max:
            max = num
    return max
print(max_num(arr))
```

```python
def find_max(arr):
    max_num = arr[0]
    for num in arr[1:]:
        if num > max_num:
            max_num = num
    return max_num

print(find_max(arr))  # 9
```

```python
def find_duplicates(arr):
    seen =set()
    duplicates =set()
    for i in arr:
        if i in seen:
            duplicates.add(i)
        else : seen.add(i)
    return duplicates

print(find_duplicates(arr))
```

```python
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1   # need a bigger sum, move left pointer up
        else:
            right -= 1  # need a smaller sum, move right pointer down
    
    return []

arr = [1, 3, 5, 7, 9, 11]
print(two_sum_sorted(arr, 12))
```
