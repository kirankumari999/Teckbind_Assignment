```python
#Traversal - Visiting each elements one by one with index or with index
```

```python
arr = [1, 2, 3, 4, 5]

for x in arr:
    print(x)
```

```python
# or with index access:
for i in range(len(arr)):
    print(i, arr[i])
```

```python
for i,x in enumerate(arr):
    print(i, arr[i])
```

```python
#Mapping
```

```python
def double(x):
    return x*2
   
```

```python
arr  = [1,2,3,4,5]
data = [double(x) for x in arr]
```

```python
data1 = list(map(lambda x: x*2, arr))
```

```python
filter = []
for i in arr:
    if i <5:
        filter.append(i)
        print(filter)
```

```python
arr  = [1,2,3,4,5]
filter2= list[filter(lambda x: x > 5, arr)]
```

```python
arr = [1, 2, 3, 4]

total = sum(arr)     
largest = max(arr)   
smallest = min(arr)
print("total",total)
```

```python
from functools import reduce
total = reduce(lambda acc, x: acc + x, arr, 0)
print(total)
```

```python
arr = [4, 2, 5, 1, 3]

arr.sort()                     # [1, 2, 3, 4, 5] - in place, modifies original
arr.sort(reverse=True)          # [5, 4, 3, 2, 1] - descending
arr.sort(key=lambda x: -x)       # custom sort rule

new_arr = sorted(arr)  
```
