```python
def two_sum(arr, target):                                               
    seen = {}  # value -> index
    
    for i, num in enumerate(arr):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    
    return []  # no pair found

arr = [2, 7, 11, 15]
target = 9
print(two_sum(arr, target))  # [0, 1] → arr[0]+arr[1] = 2+7 = 9  ## O(n) Hash concept
```

```python
def two_sum_brute(arr, target):    ##brute force
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] + arr[j] == target:
                return [i, j]
    return []

print(two_sum_brute(arr, target)) ## O(n2)
```

```python
def max_sum_pair(arr):
    arr_sorted = sorted(arr, reverse=True)
    return arr_sorted[0] + arr_sorted[1]

arr = [4, 1, 9, 7, 3, 9, 2]
print(max_sum_pair(arr))
```
